```python
import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
from scipy.optimize import milp, LinearConstraint, Bounds
from scipy.sparse import dok_matrix

pd.options.display.float_format = lambda x: f"{x:,.2f}"

RANDOM_STATE = 42
TURNAROUND_BUFFER = 15
MAX_SAMPLE = 150_000


def hhmm_to_minutes(value: float) -> int:
    value = int(value)
    hours = value // 100
    minutes = value % 100
    return hours * 60 + minutes

def add_midnight_if_needed(dep_minutes: int, arr_value: float) -> int:
    raw_arr = hhmm_to_minutes(arr_value)
    if raw_arr < dep_minutes:
        raw_arr += 24 * 60
    return raw_arr

def minutes_to_clock(minutes_total: float) -> str:
    total = int(round(minutes_total))
    day_offset, remainder = divmod(total, 24 * 60)
    hours, minutes = divmod(remainder, 60)
    clock = f"{hours:02d}:{minutes:02d}"
    return clock if day_offset == 0 else f"{clock} (+{day_offset}d)"
```

```python
#Este modelo es referencial, el target y variables pueden estar erróneas. Analiza y mejora el modelo según tu criterio y lo que se indica en el contexto inicial.
RANDOM_STATE = 42
TURNAROUND_BUFFER = 15
MAX_SAMPLE = 150_000
feature_cols = [
    'OP_CARRIER',
    'ORIGIN',
    'DEST',
    'CRS_DEP_TIME',
    'DISTANCE'
    ]

df_final['FL_DATE'] = pd.to_datetime(df_final['FL_DATE'])
df_final['MONTH'] = df_final['FL_DATE'].dt.month
df_final = df_final[df_final["ARR_DELAY"].isna() == False]
model_df = df_final.sample(n=min(len(df_final), MAX_SAMPLE), random_state=RANDOM_STATE)

X = model_df[feature_cols]
y = model_df['ARR_DELAY']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=RANDOM_STATE)
preprocessor = ColumnTransformer([
    ('categorical', OneHotEncoder(handle_unknown='ignore'), ['OP_CARRIER','ORIGIN','DEST']),
    ('numeric', SimpleImputer(strategy='mean'), ['CRS_DEP_TIME','DISTANCE'])
])

delay_model = Pipeline([
    ('prep', preprocessor),
    ('model', RandomForestRegressor(n_estimators=160, max_depth=4, random_state=RANDOM_STATE, n_jobs=-1))
])

delay_model.fit(X_train, y_train)

mae = mean_absolute_error(y_test, delay_model.predict(X_test))
print(f"MAE del modelo de retrasos: {mae:,.2f} minutos")
```

### Formulación matemática del scheduling
Sea $F$ el conjunto de vuelos del escenario y $C$ el conjunto de tripulaciones disponibles. Cada arco $(i,j) \in A$ indica que el destino del vuelo $i$ coincide con el origen del vuelo $j$ y que, usando el retraso previsto $\hat r_i$, se cumple el colchón mínimo de conexión (`ready_time_i \leq dep_j`). Definimos:

- $x_{cf} = 1$ si la tripulación $c$ opera el vuelo $f$.
- $s_{cf} = 1$ si $f$ es el primer vuelo de $c$; $e_{cf} = 1$ si es el último.
- $y_{cij} = 1$ cuando $c$ vuela $i$ inmediatamente antes que $j$ (para $(i,j) \in A$).
- $u_c$ binaria que indica si se usa la tripulación $c$.
- $d_c$ minutos de servicio asignados a $c$, aproximados por la suma del tiempo bloque previsto + colchón por vuelo ($\beta_f$).
- $o_c$ minutos de horas extra (variable de holgura no negativa).

Con parámetros $\gamma_c$ (costo fijo por activar la tripulación) y $H_c$ (límite de jornada), el modelo es:

$$
\begin{align}
\min \quad & \sum_{c \in C} \gamma_c u_c + \alpha o_c \\
\text{s.a. } \quad & \sum_{c \in C} x_{cf} = 1 && \forall f \in F \\
& x_{cf} = s_{cf} + \sum_{i \in \text{Pred}(f)} y_{cif} && \forall c, f \\
& x_{cf} = e_{cf} + \sum_{j \in \text{Succ}(f)} y_{cfj} && \forall c, f \\
& \sum_{f} s_{cf} = u_c, \quad \sum_{f} e_{cf} = u_c && \forall c \\
& y_{cij} \le x_{ci}, \quad y_{cij} \le x_{cj} && \forall c, (i,j) \in A \\
& d_c = \sum_{f \in F} \beta_f x_{cf} && \forall c \\
& d_c - o_c \le H_c && \forall c \\
& x_{cf}, y_{cij}, s_{cf}, e_{cf}, u_c \in \{0,1\}, \quad d_c, o_c \ge 0
\end{align}
$$

$A$ y $\beta_f$ dependen de las predicciones: retrasos mayores reducen conexiones factibles y elevan el tiempo bloque previsto que alimenta la restricción de horas extra.

```python
scenario_ids = [1,2,4]
scenario = df_final[(df_final['FL_DATE'] == pd.Timestamp('2017-01-01')) &
                   (df_final['OP_CARRIER'] == 'AA') &
                   (df_final['OP_CARRIER_FL_NUM'].isin(scenario_ids))].copy()
if scenario.empty:
    raise ValueError('No se encontraron vuelos para el escenario seleccionado')

scenario['prediction'] = delay_model.predict(scenario[feature_cols])
scenario['flight_id'] = scenario['OP_CARRIER_FL_NUM'].astype(str) + '_' + scenario['ORIGIN'] + scenario['DEST']
scenario['dep_minutes'] = scenario['CRS_DEP_TIME'].apply(hhmm_to_minutes)
scenario['arr_sched_minutes'] = scenario.apply(lambda row: add_midnight_if_needed(row['dep_minutes'], row['CRS_ARR_TIME']), axis=1)
scenario['pred_arrival_minutes'] = scenario['arr_sched_minutes'] + scenario['prediction']
scenario['ready_time'] = scenario['pred_arrival_minutes'] + scenario['TAXI_IN'] + TURNAROUND_BUFFER
scenario['block_with_buffer'] = (scenario['pred_arrival_minutes'] - scenario['dep_minutes']) + TURNAROUND_BUFFER
scenario = scenario.sort_values('dep_minutes').reset_index(drop=True)
view = scenario[['flight_id','OP_CARRIER_FL_NUM','ORIGIN','DEST','CRS_DEP_TIME','CRS_ARR_TIME','prediction']].copy()
view['pred_arr_local'] = scenario['pred_arrival_minutes'].apply(minutes_to_clock)
view['ready_for_next'] = scenario['ready_time'].apply(minutes_to_clock)
view.rename(columns={'prediction':'pred_delay_min'}, inplace=True)
view
```

```python
HOME_BASES = None  # asigna ['LGA','DCA'] para fijar bases manualmente
REGULAR_PER_BASE = 2
RESERVE_PER_BASE = 1
REGULAR_COST = 80
RESERVE_COST = 200
REGULAR_DUTY_MIN = 8 * 60
RESERVE_DUTY_MIN = 9 * 60
UNBALANCED_TOUR_PENALTY = 120
overtime_penalty = 3.0
MILP_OPTIONS = {"time_limit": 60, "mip_rel_gap": 0.05}

def infer_bases_from_scenario(df, manual_bases=None):
    if manual_bases:
        return manual_bases
    order = (
        df.groupby('ORIGIN')
          .size()
          .sort_values(ascending=False)
          .index
          .tolist()
    )
    return order

def build_crew_pool(bases, regular_per_base=REGULAR_PER_BASE, reserve_per_base=RESERVE_PER_BASE):
    pool = []
    for base in bases:
        for k in range(regular_per_base):
            pool.append({
                'id': f"{base}_REG_{k+1}",
                'base': base,
                'max_duty': REGULAR_DUTY_MIN,
                'fixed_cost': REGULAR_COST,
            })
        for k in range(reserve_per_base):
            pool.append({
                'id': f"{base}_RSV_{k+1}",
                'base': base,
                'max_duty': RESERVE_DUTY_MIN,
                'fixed_cost': RESERVE_COST,
            })
    return pool

def optimize(scenario):
    bases = infer_bases_from_scenario(scenario, manual_bases=HOME_BASES)
    crew_pool = build_crew_pool(bases)
    crew_pool_df = pd.DataFrame(crew_pool)
    crew_pool_df

    flight_ids = scenario['flight_id'].tolist()
    flight_info = scenario.set_index('flight_id')[
        ['ORIGIN','DEST','dep_minutes','pred_arrival_minutes','ready_time','block_with_buffer','prediction']
    ].to_dict('index')

    pairs = []
    for i, fi in scenario.iterrows():
        for j, fj in scenario.iterrows():
            if fj['dep_minutes'] <= fi['dep_minutes']:
                continue
            if fi['DEST'] != fj['ORIGIN']:
                continue
            if fi['ready_time'] <= fj['dep_minutes']:
                pairs.append((fi['flight_id'], fj['flight_id']))

    preds = {f: [] for f in flight_ids}
    succs = {f: [] for f in flight_ids}
    for i, j in pairs:
        preds[j].append(i)
        succs[i].append(j)

    class VarStore:
        def __init__(self):
            self.lower = []
            self.upper = []
            self.integrality = []
            self.lookup = {}
        def add(self, key, lb=0.0, ub=1.0, kind='binary'):
            idx = len(self.lower)
            self.lookup[key] = idx
            self.lower.append(lb)
            self.upper.append(ub)
            self.integrality.append(1 if kind == 'binary' else 0)
            return idx
        def size(self):
            return len(self.lower)

    vm = VarStore()
    x_idx, s_idx, e_idx, y_idx = {}, {}, {}, {}
    for crew in crew_pool:
        c = crew['id']
        for f in flight_ids:
            x_idx[(c,f)] = vm.add(('x',c,f))
            s_idx[(c,f)] = vm.add(('s',c,f))
            e_idx[(c,f)] = vm.add(('e',c,f))
    for crew in crew_pool:
        c = crew['id']
        for i, j in pairs:
            y_idx[(c,i,j)] = vm.add(('y',c,i,j))
    use_idx = {crew['id']: vm.add(('use',crew['id'])) for crew in crew_pool}
    duty_idx = {crew['id']: vm.add(('duty',crew['id']), lb=0.0, ub=2000.0, kind='continuous') for crew in crew_pool}
    overtime_idx = {crew['id']: vm.add(('ot',crew['id']), lb=0.0, ub=2000.0, kind='continuous') for crew in crew_pool}
    return_idx = {crew['id']: vm.add(('return',crew['id'])) for crew in crew_pool}

    rows, lower, upper = [], [], []

    def add_row(coeff, low, up):
        rows.append(coeff)
        lower.append(low)
        upper.append(up)

    # Cobertura
    for f in flight_ids:
        coeff = {}
        for crew in crew_pool:
            coeff[x_idx[(crew['id'], f)]] = coeff.get(x_idx[(crew['id'], f)], 0.0) + 1.0
        add_row(coeff, 1.0, 1.0)

    # Conservación de flujo
    for crew in crew_pool:
        c = crew['id']
        for f in flight_ids:
            coeff_in = {x_idx[(c,f)]: 1.0, s_idx[(c,f)]: -1.0}
            for i in preds[f]:
                coeff_in[y_idx[(c,i,f)]] = coeff_in.get(y_idx[(c,i,f)], 0.0) - 1.0
            add_row(coeff_in, 0.0, 0.0)
            coeff_out = {x_idx[(c,f)]: 1.0, e_idx[(c,f)]: -1.0}
            for j in succs[f]:
                coeff_out[y_idx[(c,f,j)]] = coeff_out.get(y_idx[(c,f,j)], 0.0) - 1.0
            add_row(coeff_out, 0.0, 0.0)

    # Inicios y finales
    for crew in crew_pool:
        c = crew['id']
        coeff_start = {use_idx[c]: -1.0}
        coeff_end = {use_idx[c]: -1.0}
        for f in flight_ids:
            coeff_start[s_idx[(c,f)]] = coeff_start.get(s_idx[(c,f)], 0.0) + 1.0
            coeff_end[e_idx[(c,f)]] = coeff_end.get(e_idx[(c,f)], 0.0) + 1.0
        add_row(coeff_start, 0.0, 0.0)
        add_row(coeff_end, 0.0, 0.0)

    # Uso vs asignaciones
    num_flights = len(flight_ids)
    for crew in crew_pool:
        c = crew['id']
        coeff_cap = {use_idx[c]: -num_flights}
        coeff_min = {use_idx[c]: -1.0}
        for f in flight_ids:
            coeff_cap[x_idx[(c,f)]] = coeff_cap.get(x_idx[(c,f)], 0.0) + 1.0
            coeff_min[x_idx[(c,f)]] = coeff_min.get(x_idx[(c,f)], 0.0) + 1.0
        add_row(coeff_cap, -np.inf, 0.0)
        add_row(coeff_min, 0.0, np.inf)

    # Compatibilidad de arcos
    for crew in crew_pool:
        c = crew['id']
        for i, j in pairs:
            add_row({y_idx[(c,i,j)]: 1.0, x_idx[(c,i)]: -1.0}, -np.inf, 0.0)
            add_row({y_idx[(c,i,j)]: 1.0, x_idx[(c,j)]: -1.0}, -np.inf, 0.0)

    # Restricciones de base para el vuelo inicial
    for crew in crew_pool:
        c = crew['id']
        base = crew['base']
        for f in flight_ids:
            if flight_info[f]['ORIGIN'] != base:
                add_row({s_idx[(c,f)]: 1.0}, 0.0, 0.0)

    # Registrar retornos al aeropuerto base
    for crew in crew_pool:
        c = crew['id']
        base = crew['base']
        coeff_return = {return_idx[c]: -1.0}
        for f in flight_ids:
            if flight_info[f]['DEST'] == base:
                coeff_return[e_idx[(c,f)]] = coeff_return.get(e_idx[(c,f)], 0.0) + 1.0
        add_row(coeff_return, 0.0, np.inf)
        add_row({return_idx[c]: 1.0, use_idx[c]: -1.0}, -np.inf, 0.0)

    # Definición duty y horas extra
    BIG = 2000.0
    for crew in crew_pool:
        c = crew['id']
        coeff_limit = {duty_idx[c]: 1.0, use_idx[c]: -BIG}
        add_row(coeff_limit, -np.inf, 0.0)
        coeff_duty = {duty_idx[c]: 1.0}
        for f in flight_ids:
            coeff_duty[x_idx[(c,f)]] = coeff_duty.get(x_idx[(c,f)], 0.0) - flight_info[f]['block_with_buffer']
        add_row(coeff_duty, 0.0, 0.0)
        coeff_ot = {duty_idx[c]: 1.0, overtime_idx[c]: -1.0}
        add_row(coeff_ot, -np.inf, crew['max_duty'])
        coeff_ot_cap = {overtime_idx[c]: 1.0, use_idx[c]: -BIG}
        add_row(coeff_ot_cap, -np.inf, 0.0)

    A = dok_matrix((len(rows), vm.size()))
    for r, coeff in enumerate(rows):
        for idx, val in coeff.items():
            A[r, idx] = val
    constraint = LinearConstraint(A.tocsr(), np.array(lower), np.array(upper))

    objective = np.zeros(vm.size())
    for crew in crew_pool:
        c = crew['id']
        objective[use_idx[c]] = crew['fixed_cost'] + UNBALANCED_TOUR_PENALTY
        objective[overtime_idx[c]] = overtime_penalty
        objective[return_idx[c]] = -UNBALANCED_TOUR_PENALTY

    solution = milp(
        c=objective,
        constraints=[constraint],
        integrality=np.array(vm.integrality),
        bounds=Bounds(np.array(vm.lower), np.array(vm.upper)),
        options=MILP_OPTIONS
    )
    if solution.x is None:
        raise RuntimeError(f'Solver sin solución factible: {solution.message}')
    if solution.status != 0:
        print(f"Resolver terminó con estado {solution.status}: {solution.message}")
    print(f'Costo total (fijo + extra): {solution.fun:,.2f}')

    return solution

print("Escenario con predicción")
optimize(scenario)
```
